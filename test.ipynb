```python
class NormalUser :
    name : str = "Alice"

user1 = NormalUser() # 객체생성 - 생성자 __init_(self)
user1.name
```

```python
from pydantic import BaseModel

class PydanticUser(BaseModel):
    name: str

user2 = PydanticUser(name = "")
user2.name
```

```python
class UserRegister(BaseModel):
    user_id : int
    user_name : str
    email : str
    is_admin : bool = False
```

```python
user = UserRegister(
    user_id = 1,
    user_name = "동준",
    email = 'abcd1234@email.com'
)
print(user)
```

```python
input_data = {
    "user_id": "1004",    # 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
    "user_name": "codemaster",
    "email": "code@example.com"
    # is_admin은 생략되었으므로 기본값 False가 적용됩니다.
}
```

```python
user = UserRegister(**input_data)
print("생성된 객체:", user)
print("유저 ID 타입:", type(user.user_id)) # <class 'int'> 출력
```

```python
user_dict = user.model_dump()
user_json = user.model_dump_json()
print("딕셔너리 변환:", user_dict)
print("JSON 문자열 변환:", user_json)
```
